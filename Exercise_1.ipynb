```python
# EXERCISE 1: MULTIPLE LINEAR REGRESSION

import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_squared_error

# Given Data
data = {
    'x1': [2, 5, 8, 12, 3, 6, 9, 14, 4, 7, 10, 15],
    'x2': [8, 15, 20, 30, 10, 18, 22, 35, 12, 19, 25, 38],
    'y':  [9, 18, 25, 38, 11, 21, 27, 44, 14, 23, 31, 47]
}

df = pd.DataFrame(data)

# Independent and dependent variables
X = df[['x1', 'x2']]
y = df['y']

# Create and train the regression model
model = LinearRegression()
model.fit(X, y)

# Display the regression coefficients
print("INTERCEPT (theta_0):", model.intercept_)
print("COEFFICIENT (theta_1):", model.coef_[0])
print("COEFFICIENT (theta_2):", model.coef_[1])

print("\nMULTIPLE LINEAR REGRESSION EQUATION:")
print(f"y = {model.intercept_:.4f} + "
      f"{model.coef_[0]:.4f}x1 + "
      f"{model.coef_[1]:.4f}x2")

# Predict ride price for x1 = 8 and x2 = 20
new_ride = pd.DataFrame({'x1': [8], 'x2': [20]})
predicted_price = model.predict(new_ride)

print("\nPREDICTED RIDE PRICE:")
print(f"${predicted_price[0]:.2f}")

# Make predictions for all given data
df['Predicted Price'] = model.predict(X)

print("\nLINEAR REGRESSION RESULTS:")
print(df.round(4))

# Evaluate the regression model
r2 = r2_score(y, model.predict(X))
mse = mean_squared_error(y, model.predict(X))

print("\nMODEL EVALUATION:")
print(f"R-squared: {r2:.6f}")
print(f"Mean Squared Error: {mse:.6f}")
```
